# BP5 Gate 5 — Decision Layer & Reporting (Prioritized Root-Cause Report)
**Customer360 Navigator Enterprise Suite — Root Cause & Driver Analytics**

## Why this gate is a prioritized report, not a per-complaint decision record

Master Plan Section 8's generic Gate 5 is "Decision/GenAI Layer & Reporting." BP3's own
Gate 5 (already delivered) answers that with a per-complaint decision record: one real
complaint in, one SHAP-explained decision + reason code out — because BP3 is a per-instance
deployed classifier. **BP5 is not that kind of BP.** BP5 Gate 1's own policy.json frames it
as population-level association testing across two real outcome fields, with no per-instance
decision anywhere in its own gate plan. Forcing BP3's per-complaint pattern onto BP5 would
manufacture a per-instance framing Gate 1 never asked for.

What Gate 5 does instead: it reads Gate 3's and Gate 4's own real, already-saved artifacts —
never re-deriving a single statistic — and synthesizes them into one **prioritized,
reason-coded root-cause report per outcome**: which real fields/categories are most strongly
associated with each outcome, ranked and cited back to their real source artifact, alongside
the real champion-validation context needed to judge how much to trust the SHAP-based
findings.

## What this gate does

For each of BP5's two real outcomes (`outcome_1_intervention_required`,
`outcome_2_timely_response_failure`), reading ONLY Gate 3/Gate 4's own real saved files:

1. **Field-level ranking** (Section 5) — BP5's 5 real non-control candidate fields
   (`Product`, `Sub-product`, `Issue`, `Sub-issue`, `Submitted via`), ranked by real Cramér's
   V from Gate 3's own `gate3_chi_square_cramers_v.csv`. `State` (control-only) is excluded,
   per Gate 1's own policy wording.
2. **Category-level findings** (Section 6) — for each of those 5 fields, the top 5 real
   categories by absolute log-odds-ratio magnitude from Gate 3's own
   `gate3_log_odds_ratio_by_category.csv`, restricted to categories with at least 30 real
   rows (`MIN_N_PER_CATEGORY = 30`). **This threshold is a real, disclosed design choice**
   (a conventional minimum-cell-count floor for a reasonably stable point estimate), never
   derived from the data, recorded in the Gate 5 config block, and printed in this
   notebook's own output.
3. **Champion SHAP feature-importance findings** (Section 7) — the top 10 real SHAP features
   per outcome from Gate 3's own saved SHAP artifacts, parsed back to their originating real
   field/category by prefix-matching (`map_shap_feature_to_driver_field()`).
4. **Company (frequency-encoded) finding** (Section 8) — `Company` is a real candidate
   process field but, being real high-cardinality, was tested in Gate 3 via a separate
   frequency-encoding + true statsmodels univariate logistic fit rather than
   chi-square/log-odds-ratio — reported here as its own finding from
   `gate3_company_freq_univariate_logistic.json`, not mixed into the Cramér's-V-ranked list
   (different method, not directly comparable on the same scale).
5. **Champion validation snapshot** (Section 9) — real held-out ROC-AUC/PR-AUC (Gate 3),
   bootstrap 95% CIs, Brier score, and confusion matrix @0.5 (Gate 4), presented as
   supporting model-quality context for how much to trust the SHAP-based findings above —
   **never itself a root-cause finding.**
6. **Barred-field governance disclosures** (Section 10) — Gate 3's own real barred-field
   diagnostics carried forward as disclosure only; this gate asserts
   `bar_relaxed_by_this_notebook is False` and **stops with a RuntimeError** if that were
   ever not the case, rather than silently carrying a relaxed bar forward.
7. **Mechanical UDAAP language review** (Section 12) — operationalizes the generic Gate 5
   compliance touchpoint ("UDAAP language review on customer-facing text") as a live,
   disclosed check on every one of this report's own generated narrative sentences, not only
   a markdown-cell assertion. **This gate refuses to save any report artifact if the check
   does not pass** (see Section 13's ordering — save happens only after the check).

Every finding in every section carries a **citation** back to its real source artifact —
BP6's own established schema (`source_bp, source_gate, source_artifact_relative_path,
source_field_or_metric, extracted_value, retrieval_timestamp_utc, verification_method`),
reused verbatim (HYPER), not a new schema invented for BP5.

## Two real bugs this gate's own sandbox verification caught and fixed before delivery

1. **A real CFPB taxonomy label false-positived the UDAAP check.** The real `Issue` category
   `'Problem caused by your funds being low'` — an actual CFPB taxonomy value, not text this
   project authored — literally contains the banned phrase "caused by." Two more real
   category labels do too (`'...due to money owed'`, `'...as a result of fraud'`). Scanning
   the raw narrative text would fail the check on real, disclosed CFPB source text this
   project never wrote and cannot change — a false positive, not a genuine violation this
   report is responsible for. **Fixed** by masking every single-quoted span (this report's
   own convention for quoting a real field/category value verbatim) out of the text before
   pattern-matching, so the check targets only this report's own authored template language.
   Every quoted real value that itself contains a banned pattern is still surfaced,
   separately and non-failingly, in `quoted_real_source_values_containing_banned_terms` —
   disclosed, never hidden (see the notebook's own printed `[UDAAP CHECK - DISCLOSURE ONLY,
   NOT A FAILURE]` line and `gate5_udaap_language_check.json`).
2. **A cross-sentence quote-parity bug in the first fix itself**, caught by re-running the
   same sandbox verification. This report's own template text uses the possessive
   `"Cramer's V"` — a single, unpaired apostrophe. When every narrative sentence was joined
   into one blob before quote-masking, that lone unpaired apostrophe shifted the odd/even
   parity of every quote character appearing later in the joined string, causing the masking
   regex to pair up the wrong openers/closers further down — in this gate's own real sandbox
   run, silently UN-masking the real `'Problem caused by your funds being low'` category
   label that fix #1 was supposed to protect. **Fixed** by checking every narrative sentence
   **separately** (`check_udaap_language_batch()`) rather than joining first — each
   sentence's own quotes are self-contained by construction, so there is no cross-sentence
   parity to break. Both fixes and their reasoning are documented inline in
   `src/models/bp5_driver_association.py`'s own docstrings, not only here.

## Prerequisite

BP5 Gate 3 **and** Gate 4 must both have been real-run by the user (this notebook reads both
gates' own real config blocks and real saved artifact files — it does not touch the Gold
layer or recompute any statistic). Per this project's standing execution-boundary rule,
Claude never runs this notebook — only the user does, in the `home_credit_env` Jupyter
kernel.

## What this gate does NOT do

- **No re-derivation.** Every number in this report is read directly from Gate 3's or Gate
  4's own already-saved, already real-run-confirmed artifact files — nothing is recomputed,
  and the Gold layer is never touched by this notebook.
- **No causal claims, anywhere.** Every finding is reported as a statistical **association**
  only — `ASSOCIATION_NOT_CAUSATION_DISCLAIMER` (BP5 Gate 1's own structural disclaimer) is
  carried into every saved report artifact verbatim, and the mechanical UDAAP check (Section
  12) is a live, disclosed guard against this report's own template language ever asserting
  otherwise.
- **No bar relaxation.** Barred-field diagnostics are disclosure-only; this gate raises if
  Gate 3's own artifact ever reports the bar as relaxed.
- **No per-complaint decision record.** See the opening section above — BP5's own
  population-level nature is why this gate is a prioritized report, not BP3's per-instance
  pattern.

## Real, disclosed design choices in this gate

- **`MIN_N_PER_CATEGORY = 30`** — a real, disclosed minimum-row-count floor for a category to
  be reported at the category-level-findings step, not derived from the data.
- **`TOP_K_FIELDS = 5`, `TOP_K_CATEGORIES_PER_FIELD = 5`, `TOP_K_SHAP_FEATURES = 10`** —
  real, disclosed reporting caps, not statistical thresholds; every real number behind them
  remains available in Gate 3's/Gate 4's own full artifacts.
- **Citation timestamps are stamped live** at the moment the user real-runs this notebook
  (`datetime.now(timezone.utc)`), never backdated or estimated.

Every real number in this report is read directly from Gate 3's and Gate 4's own real,
already Gate-2-confirmed-Gold-layer-derived, already real-run-confirmed artifacts. Nothing
is estimated, assumed, or synthesized. Every finding remains a statistical association,
never a causal claim, per BP5 Gate 1's own structural disclaimer.


```python
"""
Customer360 Navigator Enterprise Suite - BP5 Gate 5 decision layer & reporting notebook
(prioritized, reason-coded root-cause report synthesizing Gate 3 + Gate 4's own real,
already-saved artifacts). Single consolidated code cell (platform convention). Idempotent -
safe to re-run. This gate performs NO statistical re-derivation and does NOT touch the Gold
layer - it only reads Gate 3/Gate 4's own real output files and synthesizes/ranks them.
"""

import os, sys, json, warnings
from pathlib import Path

warnings.filterwarnings("ignore")


# ============================================================
# SECTION 1: Project root resolution (PROJECT_STRUCTURE_LOCKED.md rule #3)
# ============================================================
def _find_project_root() -> Path:
    marker = "PROJECT_STRUCTURE_LOCKED.md"
    env_override = os.environ.get("C360_PROJECT_ROOT")
    if env_override:
        if (Path(env_override) / marker).exists():
            return Path(env_override)
        raise RuntimeError(
            f"C360_PROJECT_ROOT is set to {env_override!r} but {marker} was not found there. "
            "Fix the environment variable rather than removing this check."
        )

    start = Path.cwd()
    cur = start
    for _ in range(8):
        if (cur / marker).exists():
            return cur
        if cur.parent == cur:
            break
        cur = cur.parent

    for depth_root, dirnames, filenames in os.walk(start):
        rel_depth = len(Path(depth_root).relative_to(start).parts)
        if rel_depth > 3:
            dirnames[:] = []
            continue
        dirnames[:] = [d for d in dirnames if not d.startswith(".")]
        if marker in filenames:
            return Path(depth_root)

    raise RuntimeError(
        f"Could not resolve PROJECT_ROOT: no {marker} found by walking up from {start}, nor by "
        "searching up to 3 levels below it. Fix: add a cell at the TOP of this notebook (before "
        "this cell runs) with:\n"
        '    import os; os.environ["C360_PROJECT_ROOT"] = r"C:\\Users\\rnand\\Documents\\'
        'Customer360_Navigator_Enterprise_Suite"\n'
        "then re-run from the top."
    )


PROJECT_ROOT = _find_project_root()
sys.path.insert(0, str(PROJECT_ROOT / "src"))
print(f"[OK] Project root resolved: {PROJECT_ROOT.name}")

# ============================================================
# SECTION 2: WARP performance configuration - FIRST, before any heavy import
# ============================================================
from utils.performance_setup import configure_performance, memory_headroom_gb

perf_summary = configure_performance(project_root=PROJECT_ROOT)
print(f"[WARP] Headroom before heavy work: {memory_headroom_gb()} GB")

# ============================================================
# SECTION 3: Heavy imports (only after WARP configuration)
# ============================================================
import pandas as pd
import yaml
from IPython.display import display

from models.bp5_driver_association import (
    ASSOCIATION_NOT_CAUSATION_DISCLAIMER,
    build_field_ranking,
    build_category_findings,
    build_shap_findings,
    build_company_frequency_finding,
    build_champion_validation_snapshot,
    build_field_narrative,
    build_category_narrative,
    build_shap_narrative,
    check_udaap_language_batch,
    UDAAP_BANNED_CAUSAL_PATTERNS,
)
from datetime import datetime, timezone

CONFIGS_DIR = PROJECT_ROOT / "configs"
ARTIFACTS_DIR = PROJECT_ROOT / "notebooks" / "bp5_root_cause_driver_analytics" / "artifacts"
ARTIFACTS_DIR.mkdir(parents=True, exist_ok=True)

BP5_CONFIG_PATH = CONFIGS_DIR / "bp5_root_cause_driver_analytics.yaml"
POLICY_JSON_PATH = ARTIFACTS_DIR / "policy.json"

REQUIRED_ARTIFACTS = {
    "chi_square": ARTIFACTS_DIR / "gate3_chi_square_cramers_v.csv",
    "log_odds": ARTIFACTS_DIR / "gate3_log_odds_ratio_by_category.csv",
    "shap_outcome_1": ARTIFACTS_DIR / "gate3_shap_top_features_outcome_1_intervention_required.csv",
    "shap_outcome_2": ARTIFACTS_DIR / "gate3_shap_top_features_outcome_2_timely_response_failure.csv",
    "champion_held_out_performance": ARTIFACTS_DIR / "gate3_champion_held_out_performance.json",
    "company_freq_univariate_logistic": ARTIFACTS_DIR / "gate3_company_freq_univariate_logistic.json",
    "barred_field_diagnostics": ARTIFACTS_DIR / "gate3_barred_field_diagnostics.json",
    "bootstrap_ci": ARTIFACTS_DIR / "gate4_bootstrap_ci.json",
    "calibration_curve": ARTIFACTS_DIR / "gate4_calibration_curve.json",
    "confusion_matrix": ARTIFACTS_DIR / "gate4_confusion_matrix.json",
}
for name, p in {**REQUIRED_ARTIFACTS, "config": BP5_CONFIG_PATH, "policy": POLICY_JSON_PATH}.items():
    if not p.exists():
        raise FileNotFoundError(f"Required Gate 3/Gate 4 real artifact not found ({name}): {p}.")

with open(POLICY_JSON_PATH, "r", encoding="utf-8") as f:
    gate1_policy = json.load(f)

with open(BP5_CONFIG_PATH, "r", encoding="utf-8") as f:
    bp5_config_text = f.read()
    bp5_config_yaml = yaml.safe_load(bp5_config_text)

gate3_marker = (
    "# --- Gate 3 (Hypothesis Testing / Regression / SHAP Association Benchmark) results "
    "(appended, idempotent overwrite) ---"
)
gate4_marker = (
    "# --- Gate 4 (Statistical Validation - Bootstrap CI / Calibration / Confusion Matrix) "
    "results (appended, idempotent overwrite) ---"
)
for marker, label in [(gate3_marker, "Gate 3"), (gate4_marker, "Gate 4")]:
    if marker not in bp5_config_text:
        raise RuntimeError(
            f"BP5 {label}'s own config block was not found - confirm {label} has been "
            "real-run before running Gate 5 (Gate 5 synthesizes both gates' own real saved "
            "artifacts, never re-derives them)."
        )
print("[OK] Confirmed BP5 Gate 3's and Gate 4's own config blocks are present.")

print(
    "\n[COMPLIANCE] ECOA/Reg B disparate-impact check: NOT APPLICABLE to BP5 (re-confirmed from "
    "Gate 1/Gate 4 - Master Plan Section 9's Regulatory Frameworks table maps ECOA/Reg B to BP1, "
    "BP2, BP3, and BP7 only). UDAAP language review on this gate's own generated customer-facing "
    "report text is performed mechanically below (Section 12), operationalizing the generic Gate "
    "5 compliance touchpoint as a live, disclosed check."
)

CANDIDATE_DRIVER_FIELDS = ["Product", "Sub-product", "Issue", "Sub-issue", "Submitted via"]
OUTCOME_1 = "outcome_1_intervention_required"
OUTCOME_2 = "outcome_2_timely_response_failure"
OUTCOMES = [OUTCOME_1, OUTCOME_2]

# ============================================================
# SECTION 4: Load Gate 3's and Gate 4's own real, already-saved artifacts (no re-derivation)
# ============================================================
chi_square_df = pd.read_csv(REQUIRED_ARTIFACTS["chi_square"])
log_odds_df = pd.read_csv(REQUIRED_ARTIFACTS["log_odds"])
shap_dfs = {
    OUTCOME_1: pd.read_csv(REQUIRED_ARTIFACTS["shap_outcome_1"]),
    OUTCOME_2: pd.read_csv(REQUIRED_ARTIFACTS["shap_outcome_2"]),
}
with open(REQUIRED_ARTIFACTS["champion_held_out_performance"], "r", encoding="utf-8") as f:
    held_out_perf = json.load(f)
with open(REQUIRED_ARTIFACTS["company_freq_univariate_logistic"], "r", encoding="utf-8") as f:
    company_freq_logistic = json.load(f)
with open(REQUIRED_ARTIFACTS["barred_field_diagnostics"], "r", encoding="utf-8") as f:
    barred_field_diagnostics = json.load(f)
with open(REQUIRED_ARTIFACTS["bootstrap_ci"], "r", encoding="utf-8") as f:
    bootstrap_ci = json.load(f)
with open(REQUIRED_ARTIFACTS["calibration_curve"], "r", encoding="utf-8") as f:
    calibration_curve = json.load(f)
with open(REQUIRED_ARTIFACTS["confusion_matrix"], "r", encoding="utf-8") as f:
    confusion_matrix = json.load(f)

print(
    f"[OK] Loaded real Gate 3/Gate 4 artifacts: chi_square rows={len(chi_square_df)}, "
    f"log_odds rows={len(log_odds_df)}, shap rows outcome_1={len(shap_dfs[OUTCOME_1])}, "
    f"shap rows outcome_2={len(shap_dfs[OUTCOME_2])}."
)

MIN_N_PER_CATEGORY = 30  # real, disclosed design choice - see build_category_findings() docstring
TOP_K_FIELDS = 5          # BP5 has exactly 5 non-control chi-square-tested fields; keep all
TOP_K_CATEGORIES_PER_FIELD = 5
TOP_K_SHAP_FEATURES = 10

# ============================================================
# SECTION 5: Field-level ranking per outcome (Gate 3 chi-square / Cramer's V)
# ============================================================
chi_square_rel_path = REQUIRED_ARTIFACTS["chi_square"].relative_to(PROJECT_ROOT).as_posix()
field_rankings = {}
for outcome in OUTCOMES:
    field_rankings[outcome] = build_field_ranking(chi_square_df, outcome, chi_square_rel_path)
    print(f"\n[FIELD RANKING] {outcome}:")
    display(pd.DataFrame(field_rankings[outcome])[["rank", "driver_field", "cramers_v", "association_strength", "p_value"]])

# ============================================================
# SECTION 6: Category-level findings per outcome (Gate 3 closed-form log-odds-ratio)
# ============================================================
log_odds_rel_path = REQUIRED_ARTIFACTS["log_odds"].relative_to(PROJECT_ROOT).as_posix()
category_findings = {}
for outcome in OUTCOMES:
    top_fields = [r["driver_field"] for r in field_rankings[outcome][:TOP_K_FIELDS]]
    category_findings[outcome] = build_category_findings(
        log_odds_df, outcome, top_fields, log_odds_rel_path,
        min_n_per_category=MIN_N_PER_CATEGORY, top_k_per_field=TOP_K_CATEGORIES_PER_FIELD,
    )
    n_total = sum(len(v) for v in category_findings[outcome].values())
    print(f"[CATEGORY FINDINGS] {outcome}: {n_total} category-level findings across {len(top_fields)} fields "
          f"(min_n_per_category={MIN_N_PER_CATEGORY}, top_k_per_field={TOP_K_CATEGORIES_PER_FIELD}).")

# ============================================================
# SECTION 7: SHAP-based champion feature-importance findings per outcome
# ============================================================
shap_rel_paths = {
    OUTCOME_1: REQUIRED_ARTIFACTS["shap_outcome_1"].relative_to(PROJECT_ROOT).as_posix(),
    OUTCOME_2: REQUIRED_ARTIFACTS["shap_outcome_2"].relative_to(PROJECT_ROOT).as_posix(),
}
shap_findings = {}
for outcome in OUTCOMES:
    shap_findings[outcome] = build_shap_findings(
        shap_dfs[outcome], outcome, CANDIDATE_DRIVER_FIELDS, shap_rel_paths[outcome],
        top_k=TOP_K_SHAP_FEATURES,
    )
    print(f"\n[SHAP FINDINGS] {outcome}:")
    display(pd.DataFrame(shap_findings[outcome])[["rank", "feature", "driver_field", "category", "mean_abs_shap"]])

# ============================================================
# SECTION 8: Company (frequency-encoded) finding per outcome
# ============================================================
company_rel_path = REQUIRED_ARTIFACTS["company_freq_univariate_logistic"].relative_to(PROJECT_ROOT).as_posix()
company_findings = {}
for outcome in OUTCOMES:
    company_findings[outcome] = build_company_frequency_finding(
        company_freq_logistic[outcome], outcome, company_rel_path,
    )
    print(
        f"[COMPANY FINDING] {outcome}: odds_ratio_per_1sd="
        f"{company_findings[outcome]['odds_ratio_per_1sd']:.6g}, p={company_findings[outcome]['p_value']:.4g}"
    )

# ============================================================
# SECTION 9: Champion validation snapshot per outcome (Gate 3 + Gate 4 - supporting context)
# ============================================================
perf_rel_path = REQUIRED_ARTIFACTS["champion_held_out_performance"].relative_to(PROJECT_ROOT).as_posix()
bootstrap_rel_path = REQUIRED_ARTIFACTS["bootstrap_ci"].relative_to(PROJECT_ROOT).as_posix()
calibration_rel_path = REQUIRED_ARTIFACTS["calibration_curve"].relative_to(PROJECT_ROOT).as_posix()
confusion_rel_path = REQUIRED_ARTIFACTS["confusion_matrix"].relative_to(PROJECT_ROOT).as_posix()

validation_snapshots = {}
for outcome in OUTCOMES:
    validation_snapshots[outcome] = build_champion_validation_snapshot(
        held_out_perf[outcome], bootstrap_ci[outcome], calibration_curve[outcome],
        confusion_matrix[outcome], outcome,
        perf_rel_path, bootstrap_rel_path, calibration_rel_path, confusion_rel_path,
    )
    print(
        f"[VALIDATION SNAPSHOT] {outcome}: held_out_roc_auc="
        f"{validation_snapshots[outcome]['held_out_roc_auc']:.6f}, brier_score="
        f"{validation_snapshots[outcome]['brier_score']:.6f}"
    )

# ============================================================
# SECTION 10: Barred-field governance disclosures (Gate 3's real diagnostics - bar NOT relaxed)
# ============================================================
barred_rel_path = REQUIRED_ARTIFACTS["barred_field_diagnostics"].relative_to(PROJECT_ROOT).as_posix()
bar_relaxed = barred_field_diagnostics.get("bar_relaxed_by_this_notebook", None)
if bar_relaxed is not True:
    print(
        f"\n[GOVERNANCE] Gate 3's own barred_field_diagnostics.json confirms "
        f"bar_relaxed_by_this_notebook={bar_relaxed!r} (must be False, never True) - the Gate 1 "
        "bars on 'Timely response?' (outcome_1), the response-duration fields (both outcomes), "
        "and 'Company response to consumer' (outcome_2) remain in place. This gate carries those "
        "real diagnostics forward as disclosure only, never as a driver finding."
    )
else:
    raise RuntimeError(
        "Gate 3's barred_field_diagnostics.json reports bar_relaxed_by_this_notebook=True - "
        "this must never happen automatically; a human governance decision is required before "
        "Gate 5 can proceed. Stopping rather than silently carrying a relaxed bar forward."
    )

# ============================================================
# SECTION 11: Assemble the prioritized root-cause report per outcome + narrative text
# ============================================================
all_narrative_sentences: list[str] = []
reports = {}
for outcome in OUTCOMES:
    narratives = {
        "field_level": [build_field_narrative(e, outcome) for e in field_rankings[outcome]],
        "category_level": {
            field: [build_category_narrative(e, outcome) for e in entries]
            for field, entries in category_findings[outcome].items()
        },
        "shap_level": [build_shap_narrative(e, outcome) for e in shap_findings[outcome]],
    }
    all_narrative_sentences.extend(narratives["field_level"])
    for sentences in narratives["category_level"].values():
        all_narrative_sentences.extend(sentences)
    all_narrative_sentences.extend(narratives["shap_level"])

    reports[outcome] = {
        "outcome_field": outcome,
        "association_not_causation_disclaimer": ASSOCIATION_NOT_CAUSATION_DISCLAIMER,
        "field_level_ranking": field_rankings[outcome],
        "category_level_findings_by_field": category_findings[outcome],
        "champion_shap_feature_importance": shap_findings[outcome],
        "company_process_field_finding": company_findings[outcome],
        "champion_validation_snapshot": validation_snapshots[outcome],
        "barred_field_governance_disclosures": barred_field_diagnostics,
        "narrative_text": narratives,
        "min_n_per_category_threshold_used": MIN_N_PER_CATEGORY,
        "top_k_fields": TOP_K_FIELDS,
        "top_k_categories_per_field": TOP_K_CATEGORIES_PER_FIELD,
        "top_k_shap_features": TOP_K_SHAP_FEATURES,
    }
    print(f"\n[ASSEMBLED] Prioritized root-cause report for {outcome}: "
          f"{len(field_rankings[outcome])} field findings, "
          f"{sum(len(v) for v in category_findings[outcome].values())} category findings, "
          f"{len(shap_findings[outcome])} SHAP findings.")

# ============================================================
# SECTION 12: Mechanical UDAAP language review on every generated narrative sentence -
# operationalizes the generic Gate 5 compliance touchpoint as a live, disclosed check. This
# gate REFUSES to save any report artifact if the check does not pass. Each sentence is
# checked SEPARATELY (check_udaap_language_batch), never by joining every sentence into one
# blob first - see that function's own docstring for the real cross-sentence quote-parity
# bug this gate's own sandbox verification caught and fixed before delivery.
# ============================================================
udaap_check_result = check_udaap_language_batch(all_narrative_sentences)
print(
    f"\n[UDAAP CHECK] Scanned {udaap_check_result['n_chars_scanned']:,} chars across "
    f"{udaap_check_result['n_sentences_scanned']} generated narrative sentences (both "
    f"outcomes) - passed={udaap_check_result['passed']}, n_sentences_failing="
    f"{udaap_check_result['n_sentences_failing']}, banned_terms_found="
    f"{udaap_check_result['banned_terms_found']}"
)
if udaap_check_result["quoted_real_source_values_containing_banned_terms"]:
    print(
        "[UDAAP CHECK - DISCLOSURE ONLY, NOT A FAILURE] Real, verbatim CFPB taxonomy values "
        "quoted in this report's narrative text happen to contain a banned phrase as literal "
        "source data (this report does not author or assert this text): "
        f"{udaap_check_result['quoted_real_source_values_containing_banned_terms']}"
    )
if not udaap_check_result["passed"]:
    raise RuntimeError(
        "UDAAP mechanical language check FAILED on this gate's own AUTHORED narrative "
        f"template text (not on quoted real source values) - failing sentences: "
        f"{udaap_check_result['failing_sentences']}. Fix the narrative-generation functions "
        "in src/models/bp5_driver_association.py rather than relaxing this check; no report "
        "artifact is saved while this check fails."
    )

# ============================================================
# SECTION 13: Save the real report artifacts (only after the UDAAP check has passed)
# ============================================================
report_paths = {}
for outcome, suffix in [(OUTCOME_1, "outcome_1_intervention_required"), (OUTCOME_2, "outcome_2_timely_response_failure")]:
    p = ARTIFACTS_DIR / f"gate5_prioritized_root_cause_report_{suffix}.json"
    with open(p, "w", encoding="utf-8") as f:
        json.dump(reports[outcome], f, indent=2, default=str)
    report_paths[outcome] = p
    print(f"[SAVED] {p.relative_to(PROJECT_ROOT)}")

udaap_check_path = ARTIFACTS_DIR / "gate5_udaap_language_check.json"
with open(udaap_check_path, "w", encoding="utf-8") as f:
    json.dump(
        {
            "checked_at_utc": datetime.now(timezone.utc).isoformat(),
            **udaap_check_result,
            "n_narrative_sentences_scanned": len(all_narrative_sentences),
            "banned_patterns_checklist": UDAAP_BANNED_CAUSAL_PATTERNS,
        },
        f, indent=2,
    )
print(f"[SAVED] {udaap_check_path.relative_to(PROJECT_ROOT)}")

# ============================================================
# SECTION 14: Write the Gate 5 config block (marker-based, order-independent - reuses
# bp1_config_sync.py unmodified, fifth BP gate this session to do so)
# ============================================================
from utils.bp1_config_sync import write_gate_block  # noqa: E402

gate5_marker = (
    "# --- Gate 5 (Decision Layer & Reporting - Prioritized Root-Cause Report) results "
    "(appended, idempotent overwrite) ---"
)
gate5_block_lines = [
    f"min_n_per_category_threshold_used: {MIN_N_PER_CATEGORY}",
    f"top_k_fields_per_outcome: {TOP_K_FIELDS}",
    f"top_k_categories_per_field: {TOP_K_CATEGORIES_PER_FIELD}",
    f"top_k_shap_features_per_outcome: {TOP_K_SHAP_FEATURES}",
    f"n_field_level_findings_outcome_1: {len(field_rankings[OUTCOME_1])}",
    f"n_field_level_findings_outcome_2: {len(field_rankings[OUTCOME_2])}",
    f"n_category_level_findings_outcome_1: {sum(len(v) for v in category_findings[OUTCOME_1].values())}",
    f"n_category_level_findings_outcome_2: {sum(len(v) for v in category_findings[OUTCOME_2].values())}",
    f"n_shap_findings_outcome_1: {len(shap_findings[OUTCOME_1])}",
    f"n_shap_findings_outcome_2: {len(shap_findings[OUTCOME_2])}",
    f"top_field_by_cramers_v_outcome_1: \"{field_rankings[OUTCOME_1][0]['driver_field']}\"",
    f"top_field_by_cramers_v_outcome_2: \"{field_rankings[OUTCOME_2][0]['driver_field']}\"",
    f"top_shap_feature_outcome_1: \"{shap_findings[OUTCOME_1][0]['feature']}\"",
    f"top_shap_feature_outcome_2: \"{shap_findings[OUTCOME_2][0]['feature']}\"",
    "udaap_language_check_passed: True",
    f"n_narrative_sentences_scanned: {len(all_narrative_sentences)}",
    "barred_fields_bar_relaxed: False",
    f'prioritized_root_cause_report_outcome_1_path: "{report_paths[OUTCOME_1].relative_to(PROJECT_ROOT).as_posix()}"',
    f'prioritized_root_cause_report_outcome_2_path: "{report_paths[OUTCOME_2].relative_to(PROJECT_ROOT).as_posix()}"',
    f'udaap_language_check_path: "{udaap_check_path.relative_to(PROJECT_ROOT).as_posix()}"',
    "ecoa_reg_b_disparate_impact_check: NOT_APPLICABLE",
    "association_not_causation_disclaimer_carried_forward: True",
]
write_gate_block(BP5_CONFIG_PATH, gate5_marker, gate5_block_lines)
print(f"[SAVED] gate5 block written to {BP5_CONFIG_PATH.relative_to(PROJECT_ROOT)}")

# ============================================================
# SECTION 15: Structural integrity checks - raise AssertionError, never silently pass
# ============================================================
checks = {
    "udaap_check_passed": udaap_check_result["passed"],
    "udaap_check_scanned_nonzero_sentences": len(all_narrative_sentences) > 0,
    "field_ranking_computed_both_outcomes": set(field_rankings.keys()) == set(OUTCOMES),
    "field_ranking_nonempty_both_outcomes": all(len(v) > 0 for v in field_rankings.values()),
    "field_ranking_sorted_desc_outcome_1": all(
        field_rankings[OUTCOME_1][i]["cramers_v"] >= field_rankings[OUTCOME_1][i + 1]["cramers_v"]
        for i in range(len(field_rankings[OUTCOME_1]) - 1)
    ),
    "field_ranking_sorted_desc_outcome_2": all(
        field_rankings[OUTCOME_2][i]["cramers_v"] >= field_rankings[OUTCOME_2][i + 1]["cramers_v"]
        for i in range(len(field_rankings[OUTCOME_2]) - 1)
    ),
    "no_control_field_in_ranking_outcome_1": all(r["driver_field"] != "State" for r in field_rankings[OUTCOME_1]),
    "no_control_field_in_ranking_outcome_2": all(r["driver_field"] != "State" for r in field_rankings[OUTCOME_2]),
    "category_findings_respect_min_n_outcome_1": all(
        e["n_rows"] >= MIN_N_PER_CATEGORY
        for entries in category_findings[OUTCOME_1].values() for e in entries
    ),
    "category_findings_respect_min_n_outcome_2": all(
        e["n_rows"] >= MIN_N_PER_CATEGORY
        for entries in category_findings[OUTCOME_2].values() for e in entries
    ),
    "shap_findings_computed_both_outcomes": set(shap_findings.keys()) == set(OUTCOMES),
    "shap_findings_sorted_desc_outcome_1": all(
        shap_findings[OUTCOME_1][i]["mean_abs_shap"] >= shap_findings[OUTCOME_1][i + 1]["mean_abs_shap"]
        for i in range(len(shap_findings[OUTCOME_1]) - 1)
    ),
    "shap_findings_sorted_desc_outcome_2": all(
        shap_findings[OUTCOME_2][i]["mean_abs_shap"] >= shap_findings[OUTCOME_2][i + 1]["mean_abs_shap"]
        for i in range(len(shap_findings[OUTCOME_2]) - 1)
    ),
    "barred_field_bar_not_relaxed": barred_field_diagnostics.get("bar_relaxed_by_this_notebook") is False,
    "every_field_finding_has_citation_outcome_1": all("citation" in r for r in field_rankings[OUTCOME_1]),
    "every_field_finding_has_citation_outcome_2": all("citation" in r for r in field_rankings[OUTCOME_2]),
    "report_json_written_outcome_1": report_paths[OUTCOME_1].exists(),
    "report_json_written_outcome_2": report_paths[OUTCOME_2].exists(),
    "udaap_check_json_written": udaap_check_path.exists(),
    "bp5_config_gate5_block_written": BP5_CONFIG_PATH.exists(),
}

print("\n=== INTEGRITY CHECKS ===")
for name, passed in checks.items():
    status = "[PASS]" if passed else "[FAIL]"
    print(f"{status} {name}")
    assert passed, f"[CHECK FAILED] {name}"

print(
    "\n[ALL CHECKS PASSED] BP5 Gate 5 complete - prioritized, reason-coded root-cause report "
    "synthesized for both real outcomes from Gate 3's and Gate 4's own real, already-saved "
    "artifacts (no re-derivation). Every finding carries a citation back to its real source "
    "artifact (BP6's own citation schema, reused). Mechanical UDAAP language check passed on "
    f"{len(all_narrative_sentences)} generated narrative sentences. Every finding remains an "
    "ASSOCIATION, never a causal claim. This is the final analytical gate in BP5's own plan; "
    "Gate 6 (Productization, Monitoring & Governance) is next per the generic Master Plan "
    "template."
)
```
